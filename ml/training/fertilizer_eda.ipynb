```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set visualization theme
sns.set_theme(style="whitegrid", palette="viridis")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
data_path = Path("../../data/raw/Fertilizer Prediction.csv")
df = pd.read_csv(data_path)
print("Dataset Shape:", df.shape)
df.head()
```

```python
# Dataset Information & Schema
df.info()
print("\nMissing Values:\n", df.isnull().sum())
print("\nDuplicate Records:", df.duplicated().sum())
```

```python
# Target Variable Distribution (Class Imbalance Check)
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x="Fertilizer Name", order=df["Fertilizer Name"].value_counts().index)
plt.title("Fertilizer Category Class Distribution")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Numerical Feature Summary Statistics
df.describe().T
```

```python
# Nutrient Concentration by Fertilizer Category
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.boxplot(ax=axes[0], data=df, x="Fertilizer Name", y="Nitrogen")
axes[0].set_title("Nitrogen (N) by Fertilizer")
axes[0].tick_params(axis="x", rotation=45)

sns.boxplot(ax=axes[1], data=df, x="Fertilizer Name", y="Phosphorous")
axes[1].set_title("Phosphorous (P) by Fertilizer")
axes[1].tick_params(axis="x", rotation=45)

sns.boxplot(ax=axes[2], data=df, x="Fertilizer Name", y="Potassium")
axes[2].set_title("Potassium (K) by Fertilizer")
axes[2].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```
